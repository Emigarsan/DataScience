### #1.2

Crea una función similar a la que se pide en el ejercicio 6 de la segunda tanda de ejercicios del workout. Es decir, una función para pintar diagramas de dispersion que tenga seis argumentos, pero ojo al cambio en el quinto:
1. Un dataframe
2. Una columna numérica del dataframe
3. Una columna numérica del dataframe diferente a la anterior.
4. Un columna categórica (Opcional, valor por defecto None)
5. Un valor para el tamaño de los puntos o una cadena, pero que puede ser una cadena de caracteres, en cuyo caso se debe interpretar como una de las columnas del dataframe cuyos valores deben controlar el tamaño
6. Un sexto argumento "escala", que permita manipular la escala del tamaño en el caso de que el quinto argumento sea una cadena. Este argumento multiplicará a los valores de la columna indicada en el parámetro anterior (así si quiero dividir por 10000, este sexto argumento tendrá que valer 1/10000)

La función debe pintar el diagrama de dispersión de la columna numérica 1, en el eje X, frente a la columna numérica 2, en el eje Y, usando un código de colores en función de la columna categórica y con el tamaño de los puntos según el argumento correspondiente. 

Si no se diera una columna categórica como argumento, el color no debe pasarse como parámetro a la función de Seaborn o Matplotlib que uses.

EXTRA OPCIONAL: Añade un argumento que controle si la función pinta una leyenda para colores y tamaños o un colorbar y una leyenda para los tamaños.

```python
def dispersion_tabla(
    df: pd.DataFrame,
    variable_numerica1: str,
    variable_numerica2: str,
    color_col: str = None,
    size_col: float | str | list | np.ndarray | pd.Series = 1,
    escala: float = 1.0,
    leyenda: bool = True
):
    df_copy = df.copy()
    fig,ax = plt.subplots(figsize=(12, 8))
    kwargs = {
        "data": df_copy,
        "x": variable_numerica1,
        "y": variable_numerica2,
        "alpha": 0.5,
        "palette": "viridis",
        "ax" : ax,
    }
    if color_col is not None:
        kwargs["hue"] = df_copy[color_col]

    if size_col is not None:
        if isinstance(size_col, (str, list, np.ndarray, pd.Series)):
            kwargs["size"] = df_copy[size_col]
            kwargs["sizes"] = (20* escala, 800* escala)

        else:
            kwargs["size"] = pd.Series([size_col] * len(df_copy))
            kwargs["sizes"] = (size_col* escala, size_col* escala)
     
        
    sns.scatterplot(**kwargs)

    if leyenda:
        handles, labels = ax.get_legend_handles_labels()
        ax.legend(
            handles=handles,
            labels=labels,
            ncol=2,
            loc="upper right",
            labelspacing=1.75,
            frameon=False)
    
    ax.set_xlabel(variable_numerica1)
    ax.set_ylabel(variable_numerica2)
    plt.tight_layout()

    return fig
```

```python
def dispersion_tabla_colorbar(
    df: pd.DataFrame,
    variable_numerica1: str,
    variable_numerica2: str,
    color_col: str = None,
    size_col: float | str | list | np.ndarray | pd.Series = 1,
    escala: float = 1.0,
    leyenda: bool = True,
    colorbar: bool = True
):
    df_copy = df.copy()
    fig, ax = plt.subplots(figsize=(12, 8))

    kwargs = {
        "data": df_copy,
        "x": variable_numerica1,
        "y": variable_numerica2,
        "alpha": 0.5,
        "ax": ax,
    }

    if color_col:
        kwargs["hue"] = color_col
        kwargs["palette"] = "viridis"

    if size_col is not None:
        if isinstance(size_col, str):
            kwargs["size"] = size_col
            kwargs["sizes"] = (20 * escala, 800 * escala)
        elif isinstance(size_col, (list, np.ndarray, pd.Series)):
            kwargs["size"] = size_col
            kwargs["sizes"] = (20 * escala, 800 * escala)
        else:
            kwargs["size"] = pd.Series([size_col] * len(df_copy))
            kwargs["sizes"] = (size_col * escala, size_col * escala)

    sns.scatterplot(**kwargs)

    # Colorbar handling when color_col is numeric
    if color_col and colorbar and pd.api.types.is_numeric_dtype(df_copy[color_col]):
        sm = plt.cm.ScalarMappable(
            cmap="viridis",
            norm=plt.Normalize(vmin=df_copy[color_col].min(), vmax=df_copy[color_col].max())
        )
        sm.set_array([])
        cbar = plt.colorbar(sm, ax=ax)
        cbar.set_label(color_col)

        if leyenda and isinstance(size_col, str):
            handles, labels = ax.get_legend_handles_labels()
            size_start_idx = None
            for i, label in enumerate(labels):
                if label == size_col:
                    size_start_idx = i
                    break

            if size_start_idx is not None:
                ax.legend(
                    handles=handles[size_start_idx:],
                    labels=labels[size_start_idx:],
                    loc="upper right",
                    frameon=False,
                    labelspacing=2
                )
            else:
                legend_obj = ax.get_legend()
                if legend_obj:
                    legend_obj.remove()
        else:
            legend_obj = ax.get_legend()
            if legend_obj:
                legend_obj.remove()

    elif leyenda:
        ax.legend(ncol=2, loc="upper right", frameon=False, labelspacing=2)
    else:
        legend_obj = ax.get_legend()
        if legend_obj:
            legend_obj.remove()

    ax.set_xlabel(variable_numerica1)
    ax.set_ylabel(variable_numerica2)
    plt.tight_layout()

    return fig
```

```python
df_titanic = pd.read_csv('data/titanic.csv')
```

1- El titanic fue un accidente con una tasa de mortalidad alta.  

```python
plt.figure(figsize=(12, 8))
conteos = df_titanic['alive'].value_counts(normalize=True)
barras = plt.bar(conteos.index, conteos.values, color=['red', 'green'])

plt.xlabel('Estado de Vida')
plt.ylabel('Proporción')
plt.title('Proporción de supervivencia en el Titanic')

for barra in barras:
    y = barra.get_height()
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        y + 0.01,
        f'{y:.1%}',
        ha='center',
        va='bottom'
    )

plt.ylim(0, conteos.max() + 0.08)
plt.show()
```

2- En esa ocasión los datos apoyan que se dio el "Las mujeres y los niños primero", pero...  

```python
plt.figure(figsize=(8, 6))
frecuencias_relativas = df_titanic.groupby('who')["alive"].value_counts(normalize=True) * 100
frecuencias_relativas = frecuencias_relativas.reset_index(name='relativa')
frecuencias_relativas = frecuencias_relativas[frecuencias_relativas['alive'] == 'yes']
frecuencias_relativas.sort_values(by='relativa', inplace=True)

ax = sns.barplot(data=frecuencias_relativas, x="alive", y="relativa", hue="who", alpha=0.8, palette="viridis")

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.1f}%', (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom', fontsize=10, xytext=(0, 3), textcoords='offset points')

plt.ylim(0, frecuencias_relativas['relativa'].max() * 1.15)
plt.ylabel('Porcentaje (%)')
plt.xlabel('')
plt.legend(title='who')
plt.tight_layout()
plt.show()
```

3- La clase en la que se viajes tuvo una influencia significativa y podría explicar...  

```python
plt.figure(figsize=(12, 8))
frecuencias = df_titanic.groupby('class')["alive"].value_counts(normalize=True)*100
frecuencias_df = frecuencias.rename("porcentaje").reset_index()

ax = sns.barplot(
    data=frecuencias_df,
    x="class",
    y="porcentaje",
    hue="alive",
    order=["First", "Second", "Third"],
    hue_order=["yes", "no"],
    palette={"yes": "seagreen", "no": "indianred"}
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

plt.title("Supervivencia (%) por clase")
plt.xlabel("Clase")
plt.ylabel("Porcentaje")
plt.ylim(0, 100)
plt.legend(title="alive")
plt.tight_layout()
plt.show()
```

4- Las diferencias de tasas de supervivencia entre mujeres y niños.    


```python
# Comparación de mujeres adultas y niños según la clase en la que viajaban
df_woman_child = df_titanic[df_titanic["who"].isin(["woman", "child"])].copy()

comp_clase = (
    df_woman_child.groupby("who")["class"]
    .value_counts(normalize=True)
    .mul(100)
    .rename("porcentaje")
    .reset_index()
)

plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=comp_clase,
    x="who",
    y="porcentaje",
    hue="class",
    order=["woman", "child"],
    hue_order=["First", "Second", "Third"],
    palette="viridis"
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=2)

plt.title("Distribución por clase: mujeres adultas vs niños")
plt.xlabel("")
plt.ylabel("Porcentaje dentro de cada grupo")
plt.ylim(0, comp_clase["porcentaje"].max() * 1.2)
plt.legend(title="Clase")
plt.tight_layout()
plt.show()
```

5- Además aunque el puerto de embarque parece influir, se podría asumir que lo fue el tipo de pasaje que embarcó en cada uno. 

```python
# Tasa de supervivencia general por ciudad de embarque
tabla_embarque = (
    df_titanic.groupby("embark_town")["alive"]
    .value_counts(normalize=True)
    .mul(100)
    .rename("porcentaje")
    .reset_index()
    .pivot(index="embark_town", columns="alive", values="porcentaje")
    .sort_index()
)

# Desglose porcentual de supervivencia por clase dentro de cada ciudad
tabla_embarque_clase = (
    df_titanic.groupby(["embark_town", "class"])["alive"]
    .value_counts(normalize=True)
    .mul(100)
    .rename("porcentaje")
    .reset_index()
)

# Nos quedamos solo con el porcentaje de supervivencia
tabla_embarque_clase = tabla_embarque_clase[tabla_embarque_clase["alive"] == "yes"]

fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=tabla_embarque_clase,
    x="embark_town",
    y="porcentaje",
    hue="class",
    order=["Cherbourg", "Queenstown", "Southampton"],
    hue_order=["First", "Second", "Third"],
    palette="viridis",
    ax=ax
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

ax.set_title("Porcentaje de supervivencia por ciudad de embarque y clase")
ax.set_xlabel("Ciudad de embarque")
ax.set_ylabel("Supervivencia (%)")
ax.legend(title="Clase")
plt.tight_layout()
plt.show()
tabla_embarque_clase
```
